# Introduction

Resembles the annealing process used to cool metals slowly to reach an ordered (low-energy) state

**Basic Idea**
- Allow **bad moves** occasionally, depending on **temperature**
- High temperature $\implies$ more bad moves allowed, shake the system out of its local minimum
- Gradually reduce temperature according to some schedule

- It is guaranteed tat if T decreased **slowly enough**, will **converge to optimal** state!
    - **Slowly enough** may mean exponentially slowly

# Pseudocode

> function simulated_annealing(problem, schedule) returns a state  
> current $\leftarrow$ problem.initial_state  
> for $t = 1$ to $\infty$ do  
> &emsp;&emsp;$T$ $\leftarrow$ schedule($t$)  
> &emsp;&emsp;if $T = 0$ then:  
> &emsp;&emsp;&emsp;&emsp;return current  
> &emsp;&emsp;successors $\leftarrow$ find successors of current  
> &emsp;&emsp;next $\leftarrow$ a randomly selected successor from successors  
> &emsp;&emsp;$\Delta E$ $\leftarrow$ next.value – current.value  
> &emsp;&emsp;if $\Delta E \gt 0$ then:  
> &emsp;&emsp;&emsp;&emsp;current $\leftarrow$ next  
> &emsp;&emsp;else:  
> &emsp;&emsp;&emsp;&emsp;current $\leftarrow$ next only with probability $e^{\Delta E / T}$
