# Introduction

The Firefly Algorithm (FA) is a nature-inspired optimization algorithm based on the **flashing behavior** of fireflies.

**Assumptions:**

- Fireflies are attracted to each other
- **Attractiveness** is proportional to **brightness**
- Less brighter firefly is attracted to the brighter firefly
- **Attractiveness decrease** as **distance** between two fireflies increase
- If brightness for both is **same**, fireflies move **randomly**
- New solutions are generated by **random walk** & **attraction** of fireflies

# Mathematically

Attractiveness (brightness) of the firefly $i$ on the firefly $j$ is based on is based on:
- Degree of the brightness of the firefly $i$
- Distance $r_ij$ between them
$$I(r) = \frac{I_s}{r^2}$$

Each firefly has a certain attractiveness value $\beta$. The attractiveness value β is relative based on the distance between fireflies:
$$\beta(r) = \beta_0 e^{- \gamma r^2}$$
Where:
- $r$: Distance between fireflies
- $\beta_0$: Firefly attractiveness value at $r = 0$
- $\gamma$: Media light absorption coefficient

**The Movement Towards Attractive Firefly**

The movement of a firefly $i$ ar position $x_i$ moving to a brighter firefly $j$ at position $x_j$ is:
$$x_i^{(t+1)} = x_i^{(t)} + \beta_0 e^{- \gamma r^2} (x_i - x_j) + \alpha \epsilon_i$$

Where:
- $\beta_0 e^{- \gamma r^2} (x_i - x_j)$: Attraction of the firefly $x_j$
- $\alpha \epsilon_i$: Randomization parameter

*Note:*
If $\beta_0 = 0$ then movement becomes random:
$$x_i^{(t+1)} = x_i^{(t)} + \alpha \epsilon_i$$

# Pseudocode

- Objective function: $f(x), \quad x =(x_{1},x_{2},...,x_{d})$
- Generate an initial population of fireflies $x_{i} \quad (i=1,2,\dots ,n)$
- Formulate light intensity $I$ so that it is associated with $f(x)$
    - (for example, for maximization problems, $I \propto f(x)$ or simply $I = f(x)$)
- Define absorption coefficient $\gamma$

> Begin  
> &emsp;&emsp;while ($t$ < MaxGeneration)  
> &emsp;&emsp;&emsp;&emsp;for $i = 1$ : $n$ (all $n$ fireflies)  
> &emsp;&emsp;&emsp;&emsp;&emsp;&emsp;for $j = 1$ : $i$ ($n$ fireflies)  
> &emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;if $I_{j} \gt I_{i}$  
> &emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;Vary attractiveness with distance $r$ via $\exp(-\gamma r)$  
> &emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;move firefly $i$ towards $j$;  
> &emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;Evaluate new solutions and update light intensity;  
> &emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;end if   
> &emsp;&emsp;&emsp;&emsp;&emsp;&emsp;end for $j$  
> &emsp;&emsp;&emsp;&emsp;end for $i$  
> &emsp;&emsp;&emsp;&emsp;Rank fireflies and find the current best;  
> &emsp;&emsp;end while  
> end
