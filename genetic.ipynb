# Introduction

- Directed search algorithms based on the mechanics of **biological evolution**
- Belongs to the family of **evolutionary algorithms (EAs)**
- Used to find optimal or near-optimal solutions in complex search spaces that are difficult to solve with traditional methods
- Mimics natural processes like reproduction, crossover, and mutation to evolve solutions over generations.

# Core Mechanism

**Key Components:**

- **Population (Generation):** A set of candidate solutions (possible solutions to the optimization problem).

- **Chromosome:** Represents a single candidate solution within the population.

- **Gene:** A single element of a chromosome, containing a value (allele) and position (locus) within the decision variable.

**Algorithm Workflow:**

1. **Initiate the Population:**
    - **Random Initialization:** The first generation is created with randomly generated candidate solutions. Population size depends on the problem and decision variable complexity.
    - **Heuristic Initialization:** Optionally, use known heuristics to initialize the population, though this may reduce diversity and is not typically recommended.
    - *Population size remains constant across all generations.*

2. **Evaluate the Fitness:**

    - A **fitness function** assigns a fitness score (probability value) to each chromosome, indicating how well it solves the objective.
    - For simple problems, the fitness function may be the objective function itself.
    - For complex problems with multiple objectives/constraints, the fitness function is designed to account for these factors.
    - **Fitness Function Characteristics:**
        - Must be fast to compute.
        - Should quantitatively measure the quality of a solution or its potential to produce fit offspring.

3. **Select Parents:**

    - Identify the fittest solutions from the population to serve as parents for the next generation.
    - Parent selection drives the algorithm toward better solutions and ensures convergence.
    - Maintains diversity to avoid premature convergence.

4. **Crossover:**

    - Generates new offspring by combining genes from parent chromosomes.
    - Randomly selects genes from each parent (e.g., uniform crossover, where genes are chosen based on a uniform distribution).
    - Offspring inherit characteristics from both parents, with the proportion of genes from each parent varying randomly.
    - Common crossover methods include single-point, double-point, and uniform crossover (uniform crossover is noted as popular).
    - The process repeats until the new generation reaches the same population size as the previous one.

5. **Mutation:**

    - Introduces random changes to genes in offspring to maintain diversity and prevent premature convergence.
    - Essential for exploration of the search space, unlike crossover, which focuses on exploitation of known solutions.
    - Helps introduce new characteristics not present in the parent chromosomes.


6. **Criteria for Termination:**

    - The algorithm stops when it converges or meets predefined criteria:
        - No significant fitness improvement over $X$ iterations.
        - The objective function reaches an optimal or predefined value.
        - The maximum number of generations is reached.

```python
import random

POP_SIZE = 500
MUT_RATE = 0.1
TARGET = 'rayan ali'
GENES = ' abcdefghijklmnopqrstuvwxyz'

def initialize_pop(TARGET):
    population = list()
    tar_len = len(TARGET)

    for i in range(POP_SIZE):
        temp = list()
        for j in range(tar_len):
            temp.append(random.choice(GENES))
        population.append(temp)

    return population

def crossover(selected_chromo, CHROMO_LEN, population):
    offspring_cross = []
    for i in range(int(POP_SIZE)):
        parent1 = random.choice(selected_chromo)
        parent2 = random.choice(population[:int(POP_SIZE*50)])

        p1 = parent1[0]
        p2 = parent2[0]

        crossover_point = random.randint(1, CHROMO_LEN-1)
        child =  p1[:crossover_point] + p2[crossover_point:]
        offspring_cross.extend([child])
    return offspring_cross

def mutate(offspring, MUT_RATE):
    mutated_offspring = []

    for arr in offspring:
        for i in range(len(arr)):
            if random.random() < MUT_RATE:
                arr[i] = random.choice(GENES)
        mutated_offspring.append(arr)
    return mutated_offspring

def selection(population, TARGET):
    sorted_chromo_pop = sorted(population, key= lambda x: x[1])
    return sorted_chromo_pop[:int(0.5*POP_SIZE)]

def fitness_cal(TARGET, chromo_from_pop):
    difference = 0
    for tar_char, chromo_char in zip(TARGET, chromo_from_pop):
        if tar_char != chromo_char:
            difference+=1
    
    return [chromo_from_pop, difference]

def replace(new_gen, population):
    for _ in range(len(population)):
        if population[_][1] > new_gen[_][1]:
          population[_][0] = new_gen[_][0]
          population[_][1] = new_gen[_][1]
    return population

def main(POP_SIZE, MUT_RATE, TARGET, GENES):
    # 1) initialize population
    initial_population = initialize_pop(TARGET)
    found = False
    population = []
    generation = 1

    # 2) Calculating the fitness for the current population
    for _ in range(len(initial_population)):
        population.append(fitness_cal(TARGET, initial_population[_]))

    # now population has 2 things, [chromosome, fitness]
    # 3) now we loop until TARGET is found
    while not found:

      # 3.1) select best people from current population
      selected = selection(population, TARGET)

      # 3.2) mate parents to make new generation
      population = sorted(population, key= lambda x:x[1])
      crossovered = crossover(selected, len(TARGET), population)
            
      # 3.3) mutating the childeren to diversfy the new generation
      mutated = mutate(crossovered, MUT_RATE)

      new_gen = []
      for _ in mutated:
          new_gen.append(fitness_cal(TARGET, _))

      # 3.4) replacement of bad population with new generation
      # we sort here first to compare the least fit population with the most fit new_gen

      population = replace(new_gen, population)

      
      if (population[0][1] == 0):
        print('Target found')
        print('String: ' + str(population[0][0]) + ' Generation: ' + str(generation) + ' Fitness: ' + str(population[0][1]))
        break
      print('String: ' + str(population[0][0]) + ' Generation: ' + str(generation) + ' Fitness: ' + str(population[0][1]))
      generation+=1

main(POP_SIZE, MUT_RATE, TARGET, GENES)
```
