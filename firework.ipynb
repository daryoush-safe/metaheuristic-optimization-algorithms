# Introduction

- A population of fireworks is initialized randomly in the search space.
- Each firework represents a potential solution to the optimization problem.

The typical firework algorithm is as follows:

1. Generate fireworks of the specified population size, and calculating the fitness value of each firework according to the fitness function.

2. Fireworks start to explode and calculate the explosion radius and number of sparks of each firework.

3. Generate explosion spark and gaussian mutation spark according to the explosion radius and the number of sparks.

4. Fireworks, explosive sparks and gaussian mutant sparks will constitute a new candidate population, and the current optimal value can be found out from this population.

5. Determine whether the termination condition is met. If the output optimal value is met, stop iteration, otherwise enter step (6).

6. Find new fireworks according to the selection rules, and enter step (2).

# Explosion Operator (Exploration)

Each firework explodes, generating a set of sparks (new candidate solutions).

firework's quality is depend on:
- Number of sparks
- Amplitude of sparks

- **Better fireworks** (with higher fitness):  
    Produce **more sparks** but with **smaller explosion amplitude** (local search).

- **Worse fireworks**:  
    Produce **fewer sparks** but with **larger explosion amplitude** (global search).

The **number of sparks** $s_i$ and explosion **amplitude** $A_i$ for the i-th firework are calculated as:
$$s_i = m \frac{(f_{max} - f_i) + \epsilon}{\sum_{j=1}^n (f_{max} - f_j) + \epsilon}$$
$$A_i = \hat{A} \frac{(f_i - f_{min}) + \epsilon}{\sum_{j=1}^n (f_j - f_{min}) + \epsilon}$$

Where:
- $m$: Maximum number of sparks
- $f_i$: Fitness of firework $i$
- $f_{max}, f_{min}$: Best and worst fitness in the population
- $\hat{A}$: Maximum explosion amplitude
- $\epsilon$: Small constant to avoid division by zero

# Spark Generation (Mutation)

## Explosion sparks

Generated within a radius $A_i$

To obtain the location of a spark:

> Initialize the location of the spark: $\vec{x}_j = \vec{x}_i$  
> $z = \text{round}(d \cdot \text{rand}(0,1))$  
> Randomly select $z$ dimensions of $\vec{x}_j$  
> Calculate the displacement: $h = A_i \cdot \text{rand}(-1,1)$  
> **for** each dimension $x^j_k \in$ {pre-selected $z$ dimensions of $\vec{x}_j$} **do**  
> &emsp;&emsp;$x^j_k = x^j_k + h$  
> &emsp;&emsp;**if** $x^j_k < x^{\min}_k$ **or** $x^j_k > x^{\max}_k$ **then**  
> &emsp;&emsp;&emsp;&emsp;map $x^j_k$ to the potential space: $x^j_k = x^{\min}_k + |x^j_k| \mod (x^{\max}_k - x^{\min}_k)$  
> &emsp;&emsp;**end if**  
> **end for**  


## Gaussian sparks

Some sparks are mutated using Gaussian noise to enhance diversity

Obtain the location of a specific spark:

> Initialize the location of the spark: $\vec{x}_j = \vec{x}_i$  
> $z = \text{round}(d \cdot \text{rand}(0,1))$  
> Randomly select $z$ dimensions of $\vec{x}_j$  
> Calculate the coefficient of Gaussian explosion:  
> &emsp;&emsp;$g = \text{Gaussian}(1, 1)$  
> **for** each dimension $x^j_k \in$ {pre-selected $z$ dimensions of $\vec{x}_j$} **do**  
> &emsp;&emsp;$x^j_k = x^j_k \cdot g$  
> &emsp;&emsp;**if** $x^j_k < x^{\min}_k$ **or** $x^j_k > x^{\max}_k$ **then**  
> &emsp;&emsp;&emsp;&emsp;map $x^j_k$ to the potential space:    
> &emsp;&emsp;&emsp;&emsp;$x^j_k = x^{\min}_k + |x^j_k| \mod (x^{\max}_k - x^{\min}_k)$  
> &emsp;&emsp;**end if**  
> **end for**
